```python
#libary
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# data
data_simple = {
    'Jam_Belajar': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'Nilai_Ujian': [50, 55, 62, 68, 70, 78, 85, 90, 95]
}
df_simple = pd.DataFrame(data_simple)
```

```python
# model
X = df_simple[['Jam_Belajar']]
y = df_simple['Nilai_Ujian']
model_lr = LinearRegression().fit(X, y)
```

```python
# Visuak
plt.scatter(X, y, color='blue')
plt.plot(X, model_lr.predict(X), color='red')
plt.title('Regresi Linear Sederhana')
plt.show()
```
