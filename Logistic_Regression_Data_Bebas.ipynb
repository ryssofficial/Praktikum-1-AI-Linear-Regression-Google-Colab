```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

#Deklarasi
data_log = {
    'Umur': [22, 25, 30, 35, 40, 45, 50, 55, 60],
    'Beli': [0, 0, 0, 0, 1, 1, 1, 1, 1]
}

df_log = pd.DataFrame(data_log)
```

```python
#proses
X_log = df_log[['Umur']]
y_log = df_log['Beli']

model_logit = LogisticRegression().fit(X_log, y_log)
```

```python
#Result
print(f"Peluang beli di umur 38: {model_logit.predict_proba([[38]])[0][1]}")
```
