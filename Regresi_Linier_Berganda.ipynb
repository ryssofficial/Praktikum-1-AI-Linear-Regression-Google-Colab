```python
#library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#Data
data_multi = {
    'Luas': [60, 80, 100, 120, 150],
    'Usia_Bangunan': [10, 5, 2, 8, 1],
    'Harga': [500, 750, 1100, 950, 1500]
}
df_multi = pd.DataFrame(data_multi)
```

```python
#model
X_m = df_multi[['Luas', 'Usia_Bangunan']]
y_m = df_multi['Harga']
model_multi = LinearRegression().fit(X_m, y_m)
```

```python
#Hasil
print(f"Koefisien: {model_multi.coef_}")
print(f"Intercept: {model_multi.intercept_}")
```
